```python
from datasets import load_dataset
import pandas as pd
import numpy as np

ds = load_dataset("Dingdong-Inc/FreshRetailNet-50K")
print(ds)                       # shows the splits and row counts

df = ds["train"].to_pandas()    # one row = one store x product x day
```

```python
print(df.shape)
print(df.columns.tolist())
df.head()
```

```python
# each cell holds 24 hourly values; this makes a (rows x 24) grid
status = np.vstack(df["hours_stock_status"].values)

# check what 0 and 1 mean on the dataset card (I believe 1 = stocked out)
print(np.unique(status))

# keep only opening hours 6-21 (the dataset focuses on 6:00-22:00)
S = status[:, 6:22]
```

```python
df["so_hours"] = S.sum(axis=1)                 # stockout hours that day
df["any_so"] = (df["so_hours"] > 0).astype(int)

# longest consecutive stockout run within the day
run = np.zeros(len(S))
longest = np.zeros(len(S))
for h in range(S.shape[1]):
    run = (run + 1) * S[:, h]                  # resets to 0 when in stock
    longest = np.maximum(longest, run)
df["longest_run"] = longest

# stockout still active at closing time
df["so_at_close"] = S[:, -1]
```

```python
df["dt"] = pd.to_datetime(df["dt"])
df = df.sort_values(["store_id", "product_id", "dt"])
g = df.groupby(["store_id", "product_id"])

# stockout tomorrow? (used for the recurrence probability)
df["so_next_day"] = g["any_so"].shift(-1)

# longest streak of consecutive stockout days
df["streak_id"] = (df["any_so"] != g["any_so"].shift()).cumsum()
df["streak_len"] = df.groupby("streak_id").cumcount() + 1
df["streak_len"] = df["streak_len"] * df["any_so"]

features = g.agg(
    freq_days=("any_so", "mean"),              # share of days with a stockout
    avg_so_hours=("so_hours", "mean"),
    avg_longest_run=("longest_run", "mean"),
    close_share=("so_at_close", "mean"),
    max_streak_days=("streak_len", "max"),
).reset_index()

# P(stockout tomorrow | stockout today)
rec = (df[df["any_so"] == 1]
       .groupby(["store_id", "product_id"])["so_next_day"].mean()
       .rename("p_repeat").reset_index())
features = features.merge(rec, on=["store_id", "product_id"], how="left")

features.describe()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

cols = ["freq_days", "avg_so_hours", "avg_longest_run", "max_streak_days", "p_repeat"]
X = StandardScaler().fit_transform(features[cols].fillna(0))

km = KMeans(n_clusters=4, random_state=42, n_init=10).fit(X)
features["pattern"] = km.labels_

print(features.groupby("pattern")[cols].mean().round(2))
```

```python

import numpy as np
import pandas as pd

# Convert 24-hour status lists to an array
status = np.stack(
    df["hours_stock_status"].to_numpy()
)

print("Status shape:", status.shape)
print("Unique values:", np.unique(status))

# Assumption: 0 = stockout
out = (status == 0)

# Identify the beginning of a stockout episode
previous_out = np.concatenate([
    np.zeros((len(out), 1), dtype=bool),
    out[:, :-1]
], axis=1)

starts = out & ~previous_out

# Daily metrics
df["daily_stockout_hours"] = out.sum(axis=1)
df["daily_stockout_starts"] = starts.sum(axis=1)
df["daily_stockout_any"] = out.any(axis=1).astype(int)

print(df[[
    "daily_stockout_hours",
    "daily_stockout_starts",
    "daily_stockout_any"
]].describe())
```

```python

plt.figure(figsize=(10, 5))

sns.histplot(
    df["daily_stockout_hours"],
    bins=np.arange(26) - 0.5,
    color="steelblue"
)

plt.title("Distribution of Daily Stockout Duration")
plt.xlabel("Stockout Hours per Day")
plt.ylabel("Number of Observations")
plt.xticks(range(25))
plt.grid(axis="y", alpha=0.2)
plt.show()
```

```python

store_summary = (
    df.groupby("store_id")
    .agg(
        observations=("daily_stockout_hours", "size"),
        total_stockout_hours=("daily_stockout_hours", "sum"),
        days_with_stockouts=("daily_stockout_any", "sum")
    )
)

store_summary["stockout_hour_rate"] = (
    store_summary["total_stockout_hours"]
    / (store_summary["observations"] * 24)
)

store_summary = store_summary.sort_values(
    "stockout_hour_rate",
    ascending=False
)

display(store_summary.head(10))

plt.figure(figsize=(10, 5))
store_summary["stockout_hour_rate"].head(10).plot(
    kind="bar",
    color="coral"
)

plt.title("Top 10 Stores by Stockout Hour Rate")
plt.xlabel("Store ID")
plt.ylabel("Share of Hours Out of Stock")
plt.tight_layout()
plt.show()
```
